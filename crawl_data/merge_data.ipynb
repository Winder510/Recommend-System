```python
# Import thư viện
import pandas as pd

# Đọc file all_reviews.csv
df = pd.read_csv('../data/all_reviews.csv')

# Đếm số rating mỗi sản phẩm
ratings_per_product = df.groupby('id_product').size()

# Chuyển thành DataFrame
df_product_ratings = ratings_per_product.reset_index()
df_product_ratings.columns = ['id_product', 'ratings_count']

# Lọc sản phẩm có từ 10 rating trở lên
min_ratings = 5
filtered_products = df_product_ratings[df_product_ratings['ratings_count'] >= min_ratings]

# Hiển thị kết quả
print(f"Số sản phẩm có từ {min_ratings} rating trở lên: {len(filtered_products)}")
print("\nThống kê số rating mỗi sản phẩm (sau khi lọc):")
print(filtered_products['ratings_count'].describe())
print("\nDanh sách sản phẩm có từ 10 rating trở lên:")
print(filtered_products)

# (Tùy chọn) Thêm tên sản phẩm từ df
filtered_products_with_names = filtered_products.merge(
    df[['id_product', 'name_product']].drop_duplicates(),
    on='id_product',
    how='left'
)
print("\nDanh sách sản phẩm có tên và số rating:")
print(filtered_products_with_names[['id_product', 'name_product', 'ratings_count']])
```

```python
# Import thư viện
import pandas as pd

# Đọc file all_reviews.csv
df = pd.read_csv('../data/all_reviews.csv')

# Kiểm tra cấu trúc dữ liệu
print("Thống kê mô tả của DataFrame:")
print(df.describe(include='all'))

# Nhóm theo name_product và đếm số id_product duy nhất
product_counts = df.groupby('name_product')['id_product'].nunique().reset_index()
product_counts.columns = ['name_product', 'id_count']

# Lọc các name_product có hơn 1 id_product (tức là trùng tên nhưng khác ID)
duplicate_names = product_counts[product_counts['id_count'] > 1]

# Đếm số sản phẩm trùng tên nhưng khác ID
num_duplicate_products = len(duplicate_names)
print(f"\nSố sản phẩm có tên trùng nhưng khác id_product: {num_duplicate_products}")

# Hiển thị danh sách chi tiết
print("\nDanh sách các sản phẩm trùng tên nhưng khác id_product:")
print(duplicate_names)

# (Tùy chọn) Hiển thị chi tiết các id_product cho từng name_product trùng
print("\nChi tiết id_product cho các tên sản phẩm trùng:")
for name in duplicate_names['name_product']:
    ids = df[df['name_product'] == name]['id_product'].unique()
    print(f"Tên sản phẩm: {name}, Các id_product: {ids}")

# (Tùy chọn) Thống kê số lần xuất hiện của các tên sản phẩm trùng
print("\nThống kê số lần xuất hiện của các tên sản phẩm trùng:")
duplicate_details = df[df['name_product'].isin(duplicate_names['name_product'])][['name_product', 'id_product']].drop_duplicates()
print(duplicate_details.groupby('name_product').size().reset_index(name='count'))
```
