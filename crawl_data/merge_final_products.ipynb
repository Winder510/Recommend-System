```python
# Import thư viện cần thiết
import pandas as pd

# Đọc hai file CSV
products_df = pd.read_csv('products.csv')
products_img_df = pd.read_csv('products_img.csv')

# Gộp hai DataFrame dựa trên cột 'id_product'
merged_df = products_df.merge(products_img_df[['id_product', 'img']], 
                            on='id_product', 
                            how='left')

# Lưu DataFrame đã gộp thành file mới 'final_products.csv'
merged_df.to_csv('final_products.csv', index=False)

# Hiển thị DataFrame kết quả để kiểm tra
print("Nội dung file final_products.csv:")
print(merged_df)
```
